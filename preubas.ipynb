```python
from entorno import Palace, Rooms, Creencias_Room

## Se inicializan las variables globales de obstáculos ##
EMPTY_ROOM = "."
SOLDADO = "M"
EXIT = "E"
PRECIPICIO = "P"
CORONEL_KURTZ = "CK"
COORDENADAS_INICIALES = (0,0)

LISTA = [SOLDADO, EXIT, CORONEL_KURTZ, PRECIPICIO, PRECIPICIO]

coordenadas_precipicios = []
coordenadas_soldado = []
coordenadas_salida = []

## Dimensiones del palacio ##
DIM_X = 6
DIM_Y = 6


class Capitan_Willard:

    def __init__ (self, ubicacion:tuple, tablero:list[list[Creencias_Room]]):

        self.tablero = tablero
        
        self.ubicacion = ubicacion
        self.visitados = set()
        self.frontera = set()
        
        self.precipicios = 0 # Variable que contabiliza la cantidad de precipicios encontrados
        self.ck = False # Indica si el Capitán Willard está o no con el Coronel 
        self.granada = 1 # Cantidad de granadas que tiene el Capitán
        self.soldado_muerto = False
        self.final = False

    def __str__(self):
        
        salida=""
        for fila in self.tablero:
           salida+= " ".join(str(room) for room in fila) + "\n"
        return salida
    
    def vecinos_seguros(self, x:int, y:int, palacio:Palace)->list:
        """
        Función que retorna una lista de coordenadas de los vecios seguros de la habitación actual
        
        Params:
            - self
            - x (int): coordenada x de la casilla
            - y(int); coordenada y de la casilla
            - palacio (Palace): tablero de juego
        Returns:
            - (list): lista de coordenadas vecinas segura
        """

        seguros = []
        movimientos = [(-1,0), (1,0), (0,-1), (0,1)]

        for mx, my in movimientos:
            vx, vy = x+mx, y+my
            if 0<= vx < DIM_X and 0<= vy < DIM_Y:
                if self.tablero[vx][vy].segura:
                    seguros.append((vx,vy))

        return seguros

    def coger_coord_vecinas (self, casilla_actual:Rooms, coord_x:int, coord_y:int) ->list:
        """
        Función que guarda y retorna las coordenadas de las casillas adyacentes a la dada. 
        Args:
            casilla_actual(Rooms): habitación de la que se buscan las vecinas
            coord_x (int)
            coord_y (int)
        
        Returns:
            coordenadas_vecians (list): lista de coordenadas vecinas
        """
        coordenadas_vecinas = []

        if not casilla_actual.arriba:
            coordenadas_vecinas.append((coord_x-1, coord_y))
        if not casilla_actual.abajo:
            coordenadas_vecinas.append((coord_x+1, coord_y))
        if not casilla_actual.izquierda:
            coordenadas_vecinas.append((coord_x, coord_y-1))
        if not casilla_actual.derecha:
            coordenadas_vecinas.append((coord_x, coord_y+1))

        return coordenadas_vecinas

    def actualizar_casillas (self, casilla_actual:Rooms)->None:
        """
        Función que actualiza las creencias del agente en base a lo que se sabe de la actual. 
        Args:
            casilla_actual(Rooms): habitación de la que se buscan las vecinas
       
        Returns:
            None
        """

        coord_x, coord_y = self.ubicacion

        ## Se actualiza la casilla sobre la que estamos

        # Tengo que hacerlo en algún momento pero no estoy segura de cuando
        # self.tablero[coord_x][coord_y].aspecto = "CW" 

        habitacion = self.tablero[coord_x][coord_y]
        
        habitacion.visitada = True
        habitacion.segura = True
        habitacion.posible_soldado = False
        habitacion.posible_precipicio = False
        habitacion.aspecto = "X"
        
        # Se guardan las coordenadas de las casillas visitadas
        self.visitados.add((coord_x,coord_y))
        self.frontera.discard((coord_x,coord_y))


        if casilla_actual.casilla == CORONEL_KURTZ:
            self.ck = True

        coordenadas_vecinas = self.coger_coord_vecinas(casilla_actual, coord_x, coord_y)
        
        # Se filtran los vecinos que no hayan sido visitados y que no sean seguros (evitando sobreescrituras)
        vecinos = [self.tablero[x][y] for x, y in coordenadas_vecinas 
                   if not self.tablero[x][y].visitada]
                            

        # Actualización debido a ausencia de los preceptos
        if not casilla_actual.brisa:
            for vecino in vecinos:
                vecino.posible_precipicio = False
                vecino.aspecto ="."

        if not casilla_actual.ronquido:
            for vecino in vecinos:
                vecino.posible_soldado = False
                vecino.aspecto ="."
        
        if not casilla_actual.brisa and not casilla_actual.ronquido:
            for vecino in vecinos:
                vecino.segura = True
                vecino.posible_soldado = False
                vecino.posible_precipicio = False
                vecino.aspecto ="."
        

        # Actualización por presencia de los preceptos, 
            # además, ha de cumplir que el vecino no se sepa aún que es una celda segura
        # if casilla_actual.brisa:
        #     for vecino in vecinos:
        #         vecino.posible_precipicio = True
        #         vecino.aspecto ="P?"

        # if casilla_actual.ronquido:
        #     for vecino in vecinos:
        #         vecino.posible_soldado = True
        #         vecino.aspecto ="M?"
    
        # En caso de que haya resplandor, se puede decir que puede estar cerca de la salida
        if casilla_actual.resplandor:
            for vecino in vecinos:
                vecino.posible_salida = True
                vecino.aspecto ="E?"

        for vx, vy in coordenadas_vecinas:
            vecino = self.tablero[vx][vy]

            if not vecino.visitada and vecino.segura:
                self.frontera.add((vx,vy))

    def deduccion_precipicio(self, palacio_completo:Palace)->None:
        """
        Función que deduce la posición de los precipicios en función del conocimiento sobre el entorno. 
        Args:
            palacio_compelto (Palace): palacio del juego con toda su información
       
        Returns:
            None
        """

        visitadas = self.visitados

        # Se estudian paso a paso las casillas vecinas de las celdas visitadas
        for x, y in visitadas:            
            casilla = palacio_completo.tablero[x][y]
            vecinos = self.coger_coord_vecinas(casilla, x, y)
            
            # En caso de que la casilla vecina tenga el precepto de brisa
            if casilla.brisa:

                # Se seleccionan los vecinos que no hayan sido visitados, que no contengan precipicios.
                vecinos = [self.tablero[i][j] for i,j in vecinos 
                   if not self.tablero[i][j].visitada 
                    and self.tablero[i][j].posible_precipicio]
                
                # En caso de que sólamente uno de los vecinos pueda contener un precipicio significa que seguramente es un precipicio, y, por lo tanto se guarda como tal
                if len(vecinos)==1 and self.precipicios < 3:
                    vecinos[0].aspecto = "P!"
                    vecinos[0].segura = False
                    vecinos[0].posible_soldado = False
                    vecinos[0].posible_precipicio = False
                    self.precipicios += 1

    def deduccion_soldado (self, palacio_completo:Palace)->None:
        """
        Función que deduce la posición del soldado en función del conocimiento sobre el entorno. 
        Args:
            palacio_compelto (Palace): palacio del juego con toda su información
       
        Returns:
            None
        """

        visitadas = self.visitados
        
        # Se estudian paso a paso las casillas vecinas de las celdas visitadas
        for x, y in visitadas:            
            casilla = palacio_completo.tablero[x][y]
            vecinos = self.coger_coord_vecinas(casilla, x, y)
            
            # En caso de que la casilla vecina tenga el precepto de ronquido
            if casilla.ronquido:
                
                # Se seleccionan los vecinos que no hayan sido visitados, que no contengan soldados.
                vecinos = [self.tablero[i][j] for i,j in vecinos 
                   if not self.tablero[i][j].visitada 
                   and self.tablero[i][j].posible_soldado]
                
                # En caso de que sólamente uno de los vecinos pueda contener un soldado significa que seguramente lo sea
                if len(vecinos)==1:
                    vecinos[0].aspecto = "S!"
                    vecinos[0].segura = False

                    # Debido a que hay un único soldado, significa que el resto no lo serán y
                    # por lo tanto se actualiza el conocimiento del resto del entorno
                    for fila in self.tablero:
                        for celda in fila:
                            if celda is not vecinos[0]:
                                celda.posible_soldado=False
```

```python
import random

VACIA = "|   |"
SEGURA = "| R |"
ISLA = "| I |"
ORILLA_FINAL ="| E |"
POSICION = "|CWCK|"

acciones = {"up": (-1,0), "down":(1,0), 
            "left":(0,-1), "right":(0,1), 
            "stay":(0,0)}
DIMX = 6
DIMY = 7

class Entorno():

    def __init__(self, tablero:list[list], inicio:tuple, salida:tuple, islas:list, corriente: dict):
        
        self.tablero = tablero
        self.inicio = inicio
        self.salida = salida
        self.islas = set(islas)
        self.corriente = corriente       
    
class Agente():

    def __init__(self, inicio:tuple, entorno, politica):
        
        self.estado = inicio
        self.entorno = entorno
        self.politica = politica

        
def transicion(estado:tuple, accion: str, entorno:Entorno)->list:
    """
    Función que devuelve una lista de las probabilidades y las coordenadas de la dirección a las que se le asocia
    Params:
        - estado(tuple): lugar en el que se encuentra el agente
        - accion(str): accion que va a tomar el agente
        - entorno(Entorno): entorno en el que trabaja
    Returns:
        - (list): lista de todas las posibilidades asociada a sus probabilidades
    """
    if es_terminal(estado, entorno):
        return [(1.0, estado)]
    
    x, y = estado
    acx, acy = acciones[accion]

    # Hacia abajo:
    abx, aby = acciones["down"]


    movimientos = []

    p_Down = entorno.corriente[y]

    # Movimiento que se quiere hacer
    mx, my = x+acx, y+acy
    if not (0<= mx <DIMX and 0<= my <DIMY) or (mx, my) in entorno.islas:
        return [(1.0, (x,y))]

    # En caso de que empuje la corriente hacia abajo
    bx, by = x+abx, y+aby
    if not (0<= bx <DIMX and 0<= by <DIMY) or (bx, by) in entorno.islas:
        bx, by = x, y
    
    if accion == "down":
        movimientos.append((1.0, (mx, my)))
    else:
        movimientos.append((p_Down, (bx, by)))
        movimientos.append((1-p_Down, (mx, my)))
    
    return movimientos


def recompensas(s:tuple, s2:tuple, a:str, entorno:Entorno)->int:
    """
    Función que calcula la recompensa de la acción realizada
    R(s, a, s').
    Params:
        - s(tuple): estado en el que se encuentra
        - s2(tuple): estado al que se dirige del cual se quiere saber la recompensa
        - a(str): acción que realiza
        - entorno(Entorono): espacio sobre el que trabaja el agente
    Returns:
        - (int): valor de la recompensa
    """

    if s2 == entorno.salida:
        return 100
    elif s2 in entorno.islas:
        return -100
    else:
        return -1

def es_terminal(estado:tuple, entorno: Entorno)->bool:
    """
    Función que devuelve un booleano en función de si
    el estado que analiza es un estado terminal o no.
    Params:
        - estado(tuple): estado que hay que estudiar
        - entorno(Entorno): entorno sobre el que se trabaja
    Returns:
        - (bool): especificando si se trata de un estado terminal ono 
    """
    if estado == entorno.salida or estado in entorno.islas:
        return True    
    return False

def value_iteration(entorno, gamma=0.9, epsilon=1e-4):
    """
    Iteración de valores:
        V(s) = max{}
    """

    estados = [(i,j) for i in range(DIMX) for j in range(DIMY)]
    # Valor de la función de tomar la acción
    V = {}
    # Mejor política a seguir en cada casilla
    politica = {}
    for estado in estados:        
        V[estado] = 0.0
        politica[estado] = None

    while True:
        delta = 0.0

        # Se recorren todos los estados
        for s in estados:

            # En caso de que se trate de un estado terminal, su mejor política es no moverse
            if es_terminal(s, entorno):
                politica[s] = None
                continue
        
            mejor_val = -float("inf")
            accion_mejor = None

            # Se exploran todas las acciones que se pueden realizar
            for a in list(acciones.keys()):
                q = 0.0
                # Se estudian las transiciones, la recompensa final que aporta condicionado por su probabilidad y el factor gamma
                for p, s2 in transicion(s, a, entorno):
                    r = recompensas(s, s2, a, entorno)
                    q += p*(r+gamma * V[s2])
                
                # En caso de que sea la acción con mayor recompensa, se almacena
                if q > mejor_val:
                    mejor_val = q
                    accion_mejor = a
            
            # Se actualizan los valores
            delta = max(delta, abs(mejor_val- V[s]))
            V[s] = mejor_val
            politica[s] = accion_mejor

        # Utilizamos delta como condición de parada, siendo delta la cantidad que cambia el valor desde su mejor valor al actual
        # En caso de que este sea menor que epsilon (un valor muy pequeño), significa que apenas hay cambio y que por lo tanto, se ha encontrado el camino óptimo.
        # Condición de parada por convergencia
        if delta < epsilon:
            break
    
    return V, politica
    

    
    
```
